```python
import time

parking = {}

def enter_vehicle(number):
    parking[number] = time.time()
    print("Vehicle entered")

def exit_vehicle(number):
    if number in parking:
        entry = parking.pop(number)
        hours = (time.time() - entry) / 3600
        fee = round(hours * 10, 2)
        print("Parking fee:", fee)
    else:
        print("Vehicle not found")

while True:
    choice = input("1.Enter 2.Exit 3.View 4.Quit: ")

    if choice == "1":
        num = input("Vehicle number: ")
        enter_vehicle(num)

    elif choice == "2":
        num = input("Vehicle number: ")
        exit_vehicle(num)

    elif choice == "3":
        print("Parked vehicles:", list(parking.keys()))

    else:
        break
```

```python
tasks = []

while True:
    choice = input("1.Add Task 2.View Tasks 3.Exit: ")

    if choice == "1":
        name = input("Task: ")
        category = input("Category: ")
        priority = input("Priority: ")
        tasks.append({"task": name, "category": category, "priority": priority})

    elif choice == "2":
        for t in tasks:
            print(t)

    else:
        break
```

```python
products = {"apple":50, "banana":30, "milk":60}
total = 0

while True:
    item = input("Enter product (or 'done'): ")

    if item == "done":
        break

    if item in products:
        qty = int(input("Quantity: "))
        cost = products[item] * qty
        total += cost
        print("Added:", cost)

print("Total bill:", total)
```

```python
seats = ["O"] * 10

while True:
    print("Seats:", seats)
    seat = int(input("Choose seat (0-9) or -1 to exit: "))

    if seat == -1:
        break

    if seats[seat] == "O":
        seats[seat] = "X"
        print("Seat booked")
    else:
        print("Seat already taken")
```

```python
students = {}

name = input("Student name: ")
marks = []

for i in range(3):
    m = int(input("Enter mark: "))
    marks.append(m)

avg = sum(marks)/len(marks)

students[name] = avg

print("Average marks:", avg)

if avg >= 90:
    grade = "A"
elif avg >= 75:
    grade = "B"
else:
    grade = "C"

print("Grade:", grade)
```

```python
freelancers = []
projects = []

name = input("Freelancer name: ")
skill = input("Skill: ")

freelancers.append({"name":name,"skill":skill})

client = input("Client name: ")
project = input("Project title: ")

projects.append({"client":client,"project":project})

print("Freelancers:", freelancers)
print("Projects:", projects)
```

```python
inventory = {}

while True:
    choice = input("1.Add Item 2.View Inventory 3.Exit: ")

    if choice == "1":
        item = input("Item name: ")
        qty = int(input("Quantity: "))
        inventory[item] = inventory.get(item,0) + qty

    elif choice == "2":
        print(inventory)

    else:
        break
```

```python
customers = []

while True:
    choice = input("1.Add Customer 2.View Customers 3.Exit: ")

    if choice == "1":
        name = input("Customer name: ")
        phone = input("Phone: ")
        customers.append({"name":name,"phone":phone})

    elif choice == "2":
        for c in customers:
            print(c)

    else:
        break
```
